```python
from tensorflow.keras.applications.resnet50 import ResNet50, preprocess_input, decode_predictions
from keras.preprocessing.sequence import pad_sequences
from keras.preprocessing import image
from keras.models import load_model, Model
import matplotlib.pyplot as plt
import pickle
import numpy as np
import warnings
warnings.filterwarnings("ignore")
import pyttsx3
engine=pyttsx3.init()
```

```python
model = load_model("./model_weights/model_9.h5")
model_temp = ResNet50(weights="imagenet", input_shape=(224,224,3))
# Create a new model, by removing the last layer (output layer of 1000 classes) fromthe resnet50
model_resnet = Model(model_temp.input, model_temp.layers[-2].output)
# Load the word_to_idx and idx_to_word from disk
with open("./storage/word_to_idx.pkl", "rb") as w2i:
    word_to_idx = pickle.load(w2i)
with open("./storage/idx_to_word.pkl", "rb") as i2w:
    idx_to_word = pickle.load(i2w)
max_len = 35
```

```python
def preprocess_image(img):
    img = image.load_img(img, target_size=(224,224))
    img = image.img_to_array(img)
    img = np.expand_dims(img, axis=0)
    img = preprocess_input(img)
    return img
```

```python
def encode_image(img):
    img = preprocess_image(img)
    feature_vector = model_resnet.predict(img)
    feature_vector = feature_vector.reshape(1, feature_vector.shape[1])
    return feature_vector
```

```python
def predict_caption(photo):
    in_text = "startseq"
    for i in range(max_len):
        sequence = [word_to_idx[w] for w in in_text.split() if w in word_to_idx]
        sequence = pad_sequences([sequence], maxlen=max_len, padding='post')
        ypred = model.predict([photo,sequence])
        ypred = ypred.argmax()
        word = idx_to_word[ypred]
        in_text+= ' ' +word
        if word =='endseq':
            break
    final_caption = in_text.split()
    final_caption = final_caption[1:-1]
    final_caption = ' '.join(final_caption)
    return final_caption
```

```python
import cv2
# Capture from device index or the name of a video file
capture_video = cv2.VideoCapture(0)
'''cam = cv2.VideoCapture("C:\\Users\\Admin\\PycharmProjects\\project_1\\openCV.mp4")'''
currentframe=0
while(capture_video.isOpened()):
# ret indicates whether or not the frame was captured successfully. If
# correct, then it's stored in the variable frame.
    ret, frame = capture_video.read()
#frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    if ret==True:
        if currentframe%100==0:
            name = './data/frame' + str(currentframe) + '.jpg'
            # writing the extracted images
            cv2.imwrite(name, frame)
            enc=encode_image(name)
            caption=predict_caption(enc)
            print(caption)
            engine.say(caption)
            engine.runAndWait()
            cv2.imshow('LIVE',frame)
            k = cv2.waitKey(100) & 0xFF
            if k == 113 or k == 27: # ord('q') -> 113, Esc -> 27
                break
                
    else:
        break
currentframe+=1
capture_video.release()
cv2.destroyAllWindows()
```
